```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, LabelEncoder,OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, classification_report)
                             
import numpy as np
from sklearn.compose import ColumnTransformer
```

```python
#read csv files **Make sure to change the file paths**

abm = pd.read_csv('./data/abm.csv') 
card = pd.read_csv('./data/card.csv')
cheque = pd.read_csv('./data/cheque.csv')
eft = pd.read_csv('./data/eft.csv')
emt = pd.read_csv('./data/emt.csv')
kyc_individual = pd.read_csv('./data/kyc_individual.csv')
kyc_industry_codes = pd.read_csv('./data/kyc_industry_codes.csv')
kyc_occupation_codes = pd.read_csv('./data/kyc_occupation_codes.csv')
kyc_smallbusiness = pd.read_csv('./data/kyc_smallbusiness.csv')
labels = pd.read_csv('./data/labels.csv')
westernunion = pd.read_csv('./data/westernunion.csv')
wire = pd.read_csv('./data/wire.csv')
```

```python
#merge the card,cheque and kyc on the customer id column
merged_dataset =  kyc_individual.merge(labels, on="customer_id", how="inner")
merged_dataset = merged_dataset.merge(card,on="customer_id",how="inner")
merged_dataset = merged_dataset.merge(cheque,on="customer_id",how="outer")
```

```python
#fill the missing amounts for cheque and card with 0
merged_dataset['amount_cad_x'] = merged_dataset['amount_cad_x'].fillna(0)
merged_dataset['amount_cad_y'] = merged_dataset['amount_cad_y'].fillna(0)
merged_dataset['total_amount'] = (merged_dataset['amount_cad_x'] + merged_dataset['amount_cad_y'])
```

```python
merged_dataset.head()
```

```python
#fill the missing countires with unknown and check for same country transaction
merged_dataset['country_x'] = merged_dataset['country_x'].fillna('Unknown')
merged_dataset['country_y'] = merged_dataset['country_y'].fillna('Unknown')
merged_dataset['same_country_transaction'] =(merged_dataset['country_x']==merged_dataset['country_y'])
```

```python
#fill the missing provinces with unknown and check for same province transaction
merged_dataset['province_x'] = merged_dataset['province_x'].fillna('Unknown')
merged_dataset['province_y'] = merged_dataset['province_y'].fillna('Unknown')
merged_dataset['same_province_transaction'] =(merged_dataset['province_x']==merged_dataset['province_y'])
```

```python
#fill the missing cities with unknown and check for same city transaction
merged_dataset['city_x'] = merged_dataset['city_x'].fillna('Unknown')
merged_dataset['city_y'] = merged_dataset['city_y'].fillna('Unknown')
merged_dataset['same_city_transaction'] =(merged_dataset['city_x']==merged_dataset['city_y'])
```

```python
#merge the credit and debit column for both card and cheque
merged_dataset['debit_credit'] = merged_dataset['debit_credit_x'].fillna(merged_dataset['debit_credit_y'])
```

```python
#merge the transaction_datetime column for both card and cheque
merged_dataset['transaction_datetime'] = merged_dataset['transaction_datetime_x'].fillna(merged_dataset['transaction_datetime_y'])
```

```python
#ddrop the individual columns both for card and cheque since we have merged them
merged_dataset = merged_dataset.drop(columns=[
'amount_cad_x', 'amount_cad_y',
'debit_credit_x', 'debit_credit_y',
'transaction_datetime_x', 'transaction_datetime_y',
'country_x', 'country_y',
'province_x', 'province_y',
'city_x', 'city_y',
'transaction_id_x', 'transaction_id_y'
])
```

```python
merged_dataset.head()
```

```python
#convert transaction_datetime ands onboard_date column to datetime objects
merged_dataset['transaction_datetime'] = merged_dataset['transaction_datetime'].astype(str)
merged_dataset['transaction_datetime'] = pd.to_datetime(merged_dataset['transaction_datetime'],dayfirst=False,format='mixed',errors='coerce')
merged_dataset['onboard_date'] = pd.to_datetime(merged_dataset['onboard_date'], errors='coerce')
```

```python
#get hour and day of transactions
merged_dataset['hour_of_day'] = merged_dataset['transaction_datetime'].dt.hour
merged_dataset['day_of_week'] = merged_dataset['transaction_datetime'].dt.dayofweek  
```

```python
#check how large a transaction is based on the income of hte person
merged_dataset['amount_vs_income'] = merged_dataset['total_amount'] / merged_dataset['income'].replace(0, np.nan)  
merged_dataset['amount_vs_income'] = merged_dataset['amount_vs_income'].fillna(0)
```

```python
#check if the transaction is in the top 5% of all transactions
merged_dataset['large_txn_flag'] = merged_dataset['total_amount'] > merged_dataset['total_amount'].quantile(0.95)
```

```python
#fill missing onboarding date with the customers first transaction
first_txn = merged_dataset['transaction_datetime'].min()
merged_dataset['onboard_date'] = merged_dataset['onboard_date'].fillna(first_txn)
```

```python
# Clean birth_date, impute missing values with median, and get their age at transaction time
merged_dataset['birth_date'] = pd.to_datetime(merged_dataset['birth_date'], errors='coerce')
median_birth = merged_dataset['birth_date'].median()
merged_dataset['birth_date'] = merged_dataset['birth_date'].fillna(median_birth)
merged_dataset['age'] = merged_dataset['transaction_datetime'].dt.year - merged_dataset['birth_date'].dt.year
```

```python
#fill missing merchant categories with unknown
merged_dataset['merchant_category'] = merged_dataset['merchant_category'].fillna('Unknown')
```

```python
merged_dataset = merged_dataset.drop(columns=[
'income', 'occupation_code',
'ecommerce_ind', 'marital_status'
])
```

```python
#drop rows with missing labels
merged_dataset = merged_dataset.dropna(subset=['label'])
```

```python
# Group features by type 
categorical_features = ['gender', 'merchant_category', 'debit_credit']
numerical_features = ['total_amount', 'hour_of_day', 'day_of_week', 'amount_vs_income', 'age']
boolean_features = ['large_txn_flag', 'same_country_transaction', 'same_province_transaction', 'same_city_transaction','debit_credit','gender']
dates = ['onboard_date', 'birth_date','transaction_datetime']
```

```python
# Encode boolean features as integers (0/1) so they can be used by ML models
for col in boolean_features:
    merged_dataset[col] = LabelEncoder().fit_transform(merged_dataset[col])
```

```python
#replace merchant category 'other' with an arbitrary number
merged_dataset['merchant_category'] = merged_dataset['merchant_category'].replace('other', 8888)
```

```python
#deop the date and customer id columns
merged_dataset = merged_dataset.drop(columns=dates)
merged_dataset = merged_dataset.drop(columns=['customer_id'])
```

```python
# merged_dataset.to_csv(r"C:\Users\ANING\Downloads\IMIBD\data\final.csv",index=False)
```

```python
#split into features and labels
X =  merged_dataset.drop(columns=['label'])
y = merged_dataset['label']
```

```python
#split between training and test labels
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
#define and train models
models = {
    "RandomForest": RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1),
    "DecisionTree": DecisionTreeClassifier(random_state=42),
    "LogisticRegression": LogisticRegression(max_iter=1000, random_state=42),
    "GradientBoosting": GradientBoostingClassifier(
        n_estimators=300, learning_rate=0.05, max_depth=3, random_state=42
    )
}

results = []
for name, model in models.items():
    print(f"\n Training {name} :")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1] if hasattr(model, "predict_proba") else None

    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    auc = roc_auc_score(y_test, y_proba) if y_proba is not None else None

    res = {
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Precision": round(precision_score(y_test, y_pred, zero_division=0), 4),
        "Recall": round(recall_score(y_test, y_pred, zero_division=0), 4),
        "Specificity": round(specificity, 4),
        "F1-Score": round(f1_score(y_test, y_pred, zero_division=0), 4),
        "AUC": round(auc, 4)
    }
    results.append(res)
    print(res)
```

```python
import seaborn as sns
from sklearn.metrics import confusion_matrix

def plot_conf_matrix(model, X_test, y_test, title):
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    labels = ['Non-Fraud', 'Fraud']

    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=labels, yticklabels=labels)
    plt.title(f'Confusion Matrix - {title}', fontsize=14)
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()

# Plot for Gradient Boosting
print("\nConfusion Matrix: Gradient Boosting")
plot_conf_matrix(models["GradientBoosting"], X_test, y_test, "Gradient Boosting")
```
